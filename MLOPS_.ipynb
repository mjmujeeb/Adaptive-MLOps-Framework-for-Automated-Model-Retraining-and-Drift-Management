```python
old_data = final_df.iloc[:700]    # past data (used for training)
new_data = final_df.iloc[700:]   # new incoming data
```

```python
import numpy as np

def calculate_psi(expected, actual, bins=10):
    expected_percents = np.histogram(expected, bins=bins)[0] / len(expected)
    actual_percents = np.histogram(actual, bins=bins)[0] / len(actual)

    psi_values = (expected_percents - actual_percents) * np.log(
        (expected_percents + 1e-6) / (actual_percents + 1e-6)
    )

    return np.sum(psi_values)
```

```python
psi_steps = calculate_psi(old_data['TotalSteps'], new_data['TotalSteps'])
psi_hr = calculate_psi(old_data['AvgHeartRate'], new_data['AvgHeartRate'])

print("PSI Steps:", psi_steps)
print("PSI Heart Rate:", psi_hr)
```

```python
def calculate_psi(expected, actual, bins=10):
    import numpy as np

    breakpoints = np.linspace(min(expected), max(expected), bins + 1)

    expected_counts = np.histogram(expected, bins=breakpoints)[0]
    actual_counts = np.histogram(actual, bins=breakpoints)[0]

    expected_perc = expected_counts / len(expected)
    actual_perc = actual_counts / len(actual)

    psi = np.sum((expected_perc - actual_perc) *
                 np.log((expected_perc + 1e-6) / (actual_perc + 1e-6)))

    return psi
```

```python
from xgboost import XGBRegressor

# Prepare old data
X_old = old_data.drop(['Calories','Id','Date'], axis=1)
y_old = old_data['Calories']

# Train old model
xgb_old = XGBRegressor()
xgb_old.fit(X_old, y_old)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

X_new = new_data.drop(['Calories','Id','Date'], axis=1)
y_new = new_data['Calories']

y_old_pred = xgb_old.predict(X_new)

old_mse = mean_squared_error(y_new, y_old_pred)
old_r2 = r2_score(y_new, y_old_pred)

print("🔹 OLD MODEL PERFORMANCE")
print("MSE:", old_mse)
print("R2 Score:", old_r2)
```

```python
y_new_pred = xgb.predict(X_new)

new_mse = mean_squared_error(y_new, y_new_pred)
new_r2 = r2_score(y_new, y_new_pred)

print("\n🔹 NEW MODEL (After Retraining)")
print("MSE:", new_mse)
print("R2 Score:", new_r2)
```

```python
if new_r2 > old_r2 and new_mse < old_mse:
    print("✅ Model retraining successful!")
    print("📈 Performance improved after retraining.")
    print(f"R2 improved from {old_r2:.2f} to {new_r2:.2f}")
    print(f"MSE reduced from {old_mse:.2f} to {new_mse:.2f}")
elif new_r2 > old_r2:
    print("⚠️ Partial improvement: R2 increased but error not reduced significantly.")
elif new_mse < old_mse:
    print("⚠️ Partial improvement: Error reduced but R2 not improved significantly.")
else:
    print("❌ Retraining did not improve the model.")
```
